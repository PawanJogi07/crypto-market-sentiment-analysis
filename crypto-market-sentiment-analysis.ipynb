```python


plt.figure(figsize=(8,5))
sns.barplot(data=df, x='classification', y='Closed PnL')
plt.title("Average Closed PnL by Market Sentiment")
plt.xticks(rotation=45)
plt.savefig('sentiment_profit.png')
plt.show()
```

```python
plt.figure(figsize=(10,5))

df.groupby('Coin')['Closed PnL'].sum().sort_values(ascending=False).plot(kind='bar')

plt.title("Total Profit by Coin")
plt.ylabel("Closed PnL")
plt.savefig('coin_profit.png')
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.boxplot(data=df, x='Side', y='Closed PnL')
plt.title("Buy vs Sell Profit")
plt.savefig('buy_sell.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Closed PnL'], bins=50)
plt.title("Distribution of Closed PnL")
plt.show()
```

```python
plt.figure(figsize=(50,35))

df.groupby('Coin')['Size USD'].sum().sort_values(ascending=False).plot(kind='bar')
plt.savefig('trading_volume.png')
plt.title("Trading Volume by Coin")
plt.show()
```

```python
numeric = df.select_dtypes(include=['int64','float64'])
plt.figure(figsize=(10,8))
sns.heatmap(numeric.corr(), annot=True, cmap='coolwarm')
plt.savefig('heatmap1.png')

plt.show()
```
